## What are arrays?

→ basically lists

&emsp;→ a Python `list` is a type of array

| | |
|---|---|
| → elements are indexed | → `arr[0]`, `arr[1]`, … |
| → array can be sliced | → `arr[start:stop:step]` |
| → variable size | → can add / remove elements from array |
| → heterogeneous | → elements can have different data types |

&emsp;→ a NumPy array (`ndarray`)

&emsp;&emsp;→ fixed size

&emsp;&emsp;→ homogeneous

## Python `list` vs NumPy `ndarray`

| | | |
|---|---|---|
| | `ndarray` | `list`|

## Integer Sizes

→ integers are stored as sequences of `bits` (`0`s and `1`s)

→ number of bits determines how large the integer can be

&emsp;4 bits  largest number  → `(1111)_2`  = `2^0+ 2^1+ 2^2+ 2^3` = `15`


&emsp;&emsp;&emsp;&emsp;&emsp;→ range is: `[0, 15]` (16 numbers)

&emsp;&emsp;&emsp;&emsp;&emsp;→ but may want `negative` numbers

&emsp;&emsp;&emsp;&emsp;&emsp;→ in that case, one bit is reserved to keep track of the sign

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;→ 3 bits  → `(111)_2`   = `2^0+ 2^1+ 2^2`= `7`

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp; -7 -6 … -1 -0 +0 +1 +2 … +6 +7

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp; → `0` does not need a sign  → `[-8, 7]`

## Vectorization

suppose we want to multiply every element of one array by the
corresponding element in another array


a = [1, 2, 3, 4]<br>b = [10, 20, 30, 40]
&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;→ `result = [10, 40, 90, 160]`

→ loop
```python
      result = []
       for iin range(4):
         result.append(a[i] * b[i])
         or
       [x * y for x, y in zip(a, b)]
```

at every loop, Python must:

&emsp;&emsp;→ lookup the operand objects

&emsp;&emsp;→ determine the types

&emsp;&emsp;→ try to perform the operation (if `a * b` does not work, it tries `b * a`)

| | |
|---|---|
| `C` does not have to do all that work | → significantly faster |



## Vectorization

NumPy implements things in such a way that

&emsp;&emsp;→ given `a` and `b` are NumPy arrays (`ndarray`)

&emsp;&emsp;→ given a supported function or operator

| | |
|---|---|
| a + b | → `add(a, b)`|
| a * b | → `multiply(a, b)` |
| a / b | → `divide(a, b)` |
| sin(a) / sin(b) | → `divide(sin(a), sin(b))` |


&emsp;&emsp;→ NumPy pushes the loop and calculations down into `C`

&emsp;&emsp;→ this is called vectorization

&emsp;&emsp;&emsp;→ these functions are called universal functions (ufunc)
